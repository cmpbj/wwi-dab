```python
pattern_schema = spark.conf.get("schema")
catalog = spark.conf.get("catalog")

spark.sql(f"use catalog `{catalog}`;")

schemas = spark.sql(f"Show schemas like '*{pattern_schema}*';")

print(pattern_schema)
```

```python
schemas.show()
```

```python
for schema in schemas.collect():
    spark.sql(f"DROP SCHEMA IF EXISTS `{schema[0]}` CASCADE;")
```
