# DLT pipeline

This Delta Live Tables (DLT) definition is executed using a pipeline defined in resources/wwi_dab.pipeline.yml.

```python
# Import DLT and src/wwi_dab
import dlt
import sys

sys.path.append(spark.conf.get("bundle.sourcePath", "."))
from pyspark.sql.functions import *
```

```python
# --- Bronze Layer ---

null_test = {"unique_sales_order_detail_id":"salesOrderDetailId is not null"}

@dlt.table(
    name="bronze_sales_order_details",
    table_properties={
        "quality": "bronze"
    }
)
@dlt.expect_all_or_fail(null_test)
def bronze_sales_order_details():
    return (
        spark.read
             .format("delta")
             .load("/Volumes/db-raw/wwi_raw/adw_raw/salesOrderDetails")
    )
```
